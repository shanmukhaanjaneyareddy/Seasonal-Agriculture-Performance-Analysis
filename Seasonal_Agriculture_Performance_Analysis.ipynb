```python
# ============================================================
# VOIS AICTE MAJOR PROJECT - PROJECT 6
# SEASONAL AGRICULTURE PERFORMANCE ANALYSIS
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

sns.set_style("whitegrid")

print("Libraries imported successfully!")


# ============================================================
# 2. LOAD DATASET
# ============================================================

file_path = '/content/seasonal_agriculture_performance_dataset.csv'

df = pd.read_csv(file_path)

print("=" * 70)
print("SEASONAL AGRICULTURE PERFORMANCE ANALYSIS")
print("=" * 70)

print("\nDataset loaded successfully!")
print("Number of Rows    :", df.shape[0])
print("Number of Columns :", df.shape[1])

print("\nFirst 5 Records:")
display(df.head())


# ============================================================
# 3. DATASET OVERVIEW
# ============================================================

print("\n" + "=" * 70)
print("DATASET OVERVIEW")
print("=" * 70)

print("\nDataset Shape:", df.shape)

print("\nColumn Names:")
for i, col in enumerate(df.columns, 1):
    print(f"{i}. {col}")

print("\nData Types:")
display(df.dtypes.to_frame("Data Type"))

print("\nStatistical Summary:")
display(df.describe().T.round(2))


# ============================================================
# 4. DATA QUALITY CHECK
# ============================================================

print("\n" + "=" * 70)
print("DATA QUALITY CHECK")
print("=" * 70)

print("\nMissing Values:")
missing = df.isnull().sum()

missing_table = pd.DataFrame({
    'Column': missing.index,
    'Missing Values': missing.values,
    'Missing Percentage': (missing.values / len(df) * 100).round(2)
})

display(missing_table[missing_table['Missing Values'] > 0])

print("\nDuplicate Rows:", df.duplicated().sum())


# ============================================================
# 5. DATA CLEANING
# ============================================================

df_clean = df.copy()

# Remove duplicate records
df_clean = df_clean.drop_duplicates()

# Numerical columns
numeric_columns = df_clean.select_dtypes(include=np.number).columns

# Categorical columns
categorical_columns = df_clean.select_dtypes(include='object').columns

# Fill numerical missing values using median
for col in numeric_columns:
    if df_clean[col].isnull().sum() > 0:
        df_clean[col] = df_clean[col].fillna(df_clean[col].median())

# Fill categorical missing values using mode
for col in categorical_columns:
    if df_clean[col].isnull().sum() > 0:
        df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])

print("\n" + "=" * 70)
print("DATA CLEANING COMPLETED")
print("=" * 70)

print("Original Rows :", len(df))
print("Final Rows    :", len(df_clean))
print("Missing Values After Cleaning :", df_clean.isnull().sum().sum())
print("Duplicate Rows After Cleaning :", df_clean.duplicated().sum())


# ============================================================
# 6. SEASONAL DISTRIBUTION
# ============================================================

season_counts = df_clean['Season'].value_counts()

print("\n" + "=" * 70)
print("FARMS BY SEASON")
print("=" * 70)

display(season_counts.to_frame("Number of Farms"))

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df_clean,
    x='Season'
)

plt.title('Number of Farms by Season')
plt.xlabel('Season')
plt.ylabel('Number of Farms')
plt.tight_layout()
plt.show()


# ============================================================
# 7. AGRICULTURAL PERFORMANCE BY SEASON
# ============================================================

season_performance = df_clean.groupby('Season').agg(
    Average_Yield=('Yield_Tonnes_Ha', 'mean'),
    Average_Production=('Production_Tonnes', 'mean'),
    Average_Revenue=('Revenue_INR', 'mean'),
    Average_Profit=('Profit_INR', 'mean')
).round(2)

print("\n" + "=" * 70)
print("AGRICULTURAL PERFORMANCE BY SEASON")
print("=" * 70)

display(season_performance)


# Average Yield by Season

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df_clean,
    x='Season',
    y='Yield_Tonnes_Ha',
    estimator='mean'
)

plt.title('Average Crop Yield by Season')
plt.xlabel('Season')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.tight_layout()
plt.show()


# ============================================================
# 8. ECONOMIC PERFORMANCE BY SEASON
# ============================================================

economic = df_clean.groupby('Season').agg(
    Average_Cost=('Total_Cost_INR', 'mean'),
    Average_Revenue=('Revenue_INR', 'mean'),
    Average_Profit=('Profit_INR', 'mean')
).round(2)

print("\n" + "=" * 70)
print("ECONOMIC PERFORMANCE BY SEASON")
print("=" * 70)

display(economic)


plt.figure(figsize=(10, 6))

economic.plot(
    kind='bar',
    figsize=(10, 6)
)

plt.title('Average Cost, Revenue and Profit by Season')
plt.xlabel('Season')
plt.ylabel('Amount (INR)')
plt.xticks(rotation=0)
plt.legend(title='Economic Measure')
plt.tight_layout()
plt.show()


# ============================================================
# 9. ENVIRONMENTAL CONDITIONS BY SEASON
# ============================================================

environment = df_clean.groupby('Season').agg(
    Average_Rainfall=('Rainfall_mm', 'mean'),
    Average_Temperature=('Avg_Temperature_C', 'mean'),
    Average_Humidity=('Humidity_pct', 'mean'),
    Average_Soil_Moisture=('Soil_Moisture_pct', 'mean')
).round(2)

print("\n" + "=" * 70)
print("ENVIRONMENTAL CONDITIONS BY SEASON")
print("=" * 70)

display(environment)


# Rainfall by Season

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df_clean,
    x='Season',
    y='Rainfall_mm',
    estimator='mean'
)

plt.title('Average Rainfall by Season')
plt.xlabel('Season')
plt.ylabel('Rainfall (mm)')
plt.tight_layout()
plt.show()


# ============================================================
# 10. RESOURCE USAGE BY SEASON
# ============================================================

resources = df_clean.groupby('Season').agg(
    Average_Fertilizer=('Fertilizer_kg_ha', 'mean'),
    Average_Water_Used=('Water_Used_m3', 'mean'),
    Average_Water_Efficiency=('Water_Efficiency_t_per_1000m3', 'mean')
).round(2)

print("\n" + "=" * 70)
print("RESOURCE USAGE BY SEASON")
print("=" * 70)

display(resources)


plt.figure(figsize=(10, 6))

resource_plot = df_clean.groupby('Season')[
    ['Fertilizer_kg_ha', 'Water_Used_m3']
].mean()

resource_plot.plot(
    kind='bar',
    figsize=(10, 6)
)

plt.title('Average Resource Usage by Season')
plt.xlabel('Season')
plt.ylabel('Average Usage')
plt.xticks(rotation=0)
plt.legend(['Fertilizer (kg/ha)', 'Water Used (m³)'])
plt.tight_layout()
plt.show()


# ============================================================
# 11. CROP PERFORMANCE ACROSS SEASONS
# ============================================================

crop_season = pd.pivot_table(
    df_clean,
    values='Yield_Tonnes_Ha',
    index='Crop',
    columns='Season',
    aggfunc='mean'
)

print("\n" + "=" * 70)
print("AVERAGE CROP YIELD BY CROP AND SEASON")
print("=" * 70)

display(crop_season.round(2))


plt.figure(figsize=(10, 6))

sns.heatmap(
    crop_season,
    annot=True,
    fmt='.2f',
    cmap='YlGnBu'
)

plt.title('Average Crop Yield by Crop and Season')
plt.xlabel('Season')
plt.ylabel('Crop')
plt.tight_layout()
plt.show()


# ============================================================
# 12. RAINFALL VS CROP YIELD
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df_clean,
    x='Rainfall_mm',
    y='Yield_Tonnes_Ha',
    hue='Season'
)

plt.title('Rainfall vs Crop Yield')
plt.xlabel('Rainfall (mm)')
plt.ylabel('Yield (Tonnes/Ha)')
plt.tight_layout()
plt.show()


# ============================================================
# 13. CORRELATION ANALYSIS
# ============================================================

correlation_columns = [
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Humidity_pct',
    'Sunlight_Hours_Day',
    'Soil_pH',
    'Soil_Moisture_pct',
    'Nitrogen_kg_ha',
    'Phosphorus_kg_ha',
    'Potassium_kg_ha',
    'Fertilizer_kg_ha',
    'Pesticide_Litre_ha',
    'Seed_Quality_Score',
    'Yield_Tonnes_Ha',
    'Production_Tonnes',
    'Market_Price_INR_Tonne',
    'Total_Cost_INR',
    'Revenue_INR',
    'Profit_INR',
    'Water_Used_m3',
    'Water_Efficiency_t_per_1000m3',
    'Disease_Pest_Risk_pct'
]

correlation = df_clean[correlation_columns].corr()

plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0
)

plt.title('Correlation Heatmap of Agricultural Variables')
plt.tight_layout()
plt.show()


# ============================================================
# 14. DISEASE AND PEST RISK BY SEASON
# ============================================================

risk = df_clean.groupby('Season')[
    'Disease_Pest_Risk_pct'
].mean().sort_values(ascending=False)

print("\n" + "=" * 70)
print("DISEASE AND PEST RISK BY SEASON")
print("=" * 70)

display(risk.to_frame('Average Disease/Pest Risk (%)'))


plt.figure(figsize=(8, 5))

risk.plot(kind='bar')

plt.title('Average Disease and Pest Risk by Season')
plt.xlabel('Season')
plt.ylabel('Risk (%)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# ============================================================
# 15. FINAL SEASONAL COMPARISON
# ============================================================

final_summary = df_clean.groupby('Season').agg(
    Farms=('Farm_ID', 'count'),
    Average_Yield=('Yield_Tonnes_Ha', 'mean'),
    Average_Production=('Production_Tonnes', 'mean'),
    Average_Revenue=('Revenue_INR', 'mean'),
    Average_Profit=('Profit_INR', 'mean'),
    Average_Water_Use=('Water_Used_m3', 'mean'),
    Average_Water_Efficiency=('Water_Efficiency_t_per_1000m3', 'mean'),
    Average_Disease_Risk=('Disease_Pest_Risk_pct', 'mean')
).round(2)

print("\n" + "=" * 70)
print("FINAL SEASONAL COMPARISON")
print("=" * 70)

display(final_summary)


# ============================================================
# 16. KEY FINDINGS
# ============================================================

best_yield_season = final_summary['Average_Yield'].idxmax()
best_profit_season = final_summary['Average_Profit'].idxmax()
best_revenue_season = final_summary['Average_Revenue'].idxmax()
best_water_efficiency_season = final_summary[
    'Average_Water_Efficiency'
].idxmax()
lowest_risk_season = final_summary[
    'Average_Disease_Risk'
].idxmin()

top_crop = (
    df_clean.groupby('Crop')['Yield_Tonnes_Ha']
    .mean()
    .idxmax()
)

top_profit_crop = (
    df_clean.groupby('Crop')['Profit_INR']
    .mean()
    .idxmax()
)

print("\n" + "=" * 70)
print("KEY FINDINGS")
print("=" * 70)

print("\nHighest Average Yield Season       :", best_yield_season)
print("Highest Average Profit Season      :", best_profit_season)
print("Highest Average Revenue Season     :", best_revenue_season)
print("Best Water Efficiency Season       :", best_water_efficiency_season)
print("Lowest Disease/Pest Risk Season    :", lowest_risk_season)
print("Top Crop by Average Yield          :", top_crop)
print("Top Crop by Average Profit         :", top_profit_crop)


# ============================================================
# 17. OVERALL PROJECT SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("PROJECT SUMMARY")
print("=" * 70)

print("\nTotal Farms Analyzed  :", df_clean['Farm_ID'].nunique())
print("Total Crops Analyzed  :", df_clean['Crop'].nunique())
print("Total States Analyzed :", df_clean['State'].nunique())
print("Total Districts       :", df_clean['District'].nunique())
print("Total Seasons         :", df_clean['Season'].nunique())

print("\nData cleaning completed successfully.")
print("Seasonal performance analysis completed successfully.")
print("Environmental analysis completed successfully.")
print("Resource usage analysis completed successfully.")
print("Economic analysis completed successfully.")
print("Relationship and correlation analysis completed successfully.")

print("\n" + "=" * 70)
print("PROJECT 6 ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 70)
```
